## 1. Strings como secuencias

Un `str` se lee por **posición** empezando en 0. `[-1]` es el último carácter. Ejemplo real de B2 con la palabra `Pradera`: `P(0) r(1) a(2) d(3) e(4) r(5) a(6)`.

```python
# EJECUTA. PREDICE cada carácter antes de ejecutar.
a = "Pradera"
print(a[0])   # PREDICE: ¿qué letra?
print(a[-1])  # PREDICE: ¿la última?
print(a[1])   # PREDICE
print(a[-2])  # PREDICE
```

```python
# COMPRUEBA
assert a[0] == "P"
assert a[-1] == "a"
assert a[1] == "r"
assert a[-2] == "r"
print("OK: lees posiciones positivas y negativas")
```

## 2. Slicing: inicio, fin (excluido) y paso

`a[inicio:fin]` incluye `inicio` pero **excluye** `fin`. `a[::-1]` recorre hacia atrás: invierte. Cada corte crea **otro** string; el original no cambia.

```python
# PREDICE cada línea antes de ejecutar.
a = "Pradera"
print(a[1:5])    # PREDICE (pista: posiciones 1, 2 y 3)
print(a[:3])     # PREDICE (desde el inicio hasta 3, sin incluir 3)
print(a[3:])     # PREDICE (desde 3 hasta el final)
print(a[::-1])   # PREDICE
print(a[5:1:-1]) # PREDICE (de 5 hacia 1, sin incluir 1)
```

```python
# COMPRUEBA
assert a[1:5] == "rade"
assert a[:3] == "Pra"
assert a[3:] == "dera"
assert a[::-1] == "aredarP"
assert a[5:1:-1] == "reda"
assert a == "Pradera"  # el original sigue igual
print("OK: dominas inicio, fin excluido y paso")
# ERROR ÚTIL: pasarse de largo también avisa.
try:
    a[20]
except IndexError as error:
    print("IndexError:", error)
```

## 4. Listas: ordenadas y mutables

Una **lista** guarda elementos en orden entre corchetes y SÍ admite `lista[i] = ...`. `len()` dice cuántos hay y `for` recorre en orden. Contraste clave: **string → inmutable · lista → mutable**.

```python
# EJECUTA. PREDICE la lista final línea a línea (ejemplo real de B2).
alumnos = ["Ana", "Luis", "Marta"]
alumnos[1] = "Leo"    # sustituye por posición (en un string esto fallaba)
alumnos.append("Nora") # añade al final
print(alumnos)         # PREDICE antes de ejecutar
```

```python
# COMPRUEBA: la MISMA lista refleja los cambios.
assert alumnos == ["Ana", "Leo", "Marta", "Nora"]
assert len(alumnos) == 4
vistos = []
for nombre in alumnos:
    print(nombre)
    vistos.append(nombre)
assert vistos == alumnos  # el for recorrió en orden
print("OK: crear, leer, modificar, medir y recorrer")
```

## 5. Tuplas: parecen listas, pero no se modifican

Una **tupla** agrupa valores estables (por ejemplo unas coordenadas) entre paréntesis. Detalle tramposo: `(5)` es solo un número; la tupla de un elemento lleva coma: `(5,)`.

```python
# EJECUTA. PREDICE el tipo de (5) y de (5,) ANTES de ejecutar.
punto = (3, 7)
solo = (5,)
no_es_tupla = (5)
print(punto)
print(type(solo))
print(type(no_es_tupla))  # PREDICE: ¿int o tuple?
```

```python
# COMPRUEBA: la coma crea la tupla.
assert type(solo) == tuple
assert type(no_es_tupla) == int
assert punto == (3, 7)
# ERROR ÚTIL: como el string, la tupla se lee pero no se sustituye.
try:
    punto[0] = 9
except TypeError as error:
    print("TypeError:", error)
assert punto == (3, 7)
print("OK: la coma distingue (5,) y la tupla no se modifica")
```

## 7. Procesamiento sencillo de texto

Las frases reales vienen mezcladas (`Python` frente a `PYTHON`). `.lower()` normaliza a minúsculas, `.split()` separa palabras e `in` pregunta si algo está. Resultado visible: lista limpia + dos preguntas.

## 8. Sets: valores únicos, sin prometer orden

Un **set** elimina repetidos automáticamente. Ganas unicidad; pierdes el orden (puede variar, no confíes en él). Necesidad real: dos grupos entregan asistentes con repetidos y quieres los nombres únicos.

```python
# EJECUTA. PREDICE cuántos únicos quedarán ANTES de ejecutar.
grupo_a = ["Ana", "Luis", "Ana"]
grupo_b = ["Luis", "Marta"]
todos = grupo_a + grupo_b
unicos = set(todos)
print(len(unicos))  # PREDICE
print(unicos)       # únicos; el orden puede variar: no lo uses
assert len(unicos) == 3
assert "Ana" in unicos and "Luis" in unicos and "Marta" in unicos
assert len(set(grupo_a)) == 2
print("OK: el set resolvió los duplicados")
```

**Objetivo + pista:** une tus dos listas de asistentes y cuenta únicos. Pista: `lista1 + lista2` junta, `set(...)` deduplica, `len(...)` cuenta. No compruebes con orden: usa `len` e `in`.

```python
# MINI-RETO 3 — cambia los nombres y predice el número de únicos.
manana = ["Leo", "Nora", "Leo"]
tarde = ["Nora", "Iris"]
unicos_reto = set(manana + tarde)
print(len(unicos_reto))
print(unicos_reto)
assert len(unicos_reto) == 3
assert "Leo" in unicos_reto
print("OK")
```

## 9. Diccionarios: clave → valor

Un **diccionario** asocia cada **clave** con su **valor** entre llaves. Se lee con `d[clave]` y se modifica igual; si la clave no existe, se crea. Caso real: inventario producto → cantidad.

```python
# EJECUTA. PREDICE cada print.
stock = {"lapiz": 4, "cuaderno": 2}
print(stock["lapiz"])  # lectura
stock["lapiz"] = 3     # modificación
stock["goma"] = 5      # clave nueva: se crea
print(stock)
assert stock == {"lapiz": 3, "cuaderno": 2, "goma": 5}
print("OK: leer, modificar y crear por clave")
```

**ERROR ÚTIL + solución.** Pedir una clave que no existe con `[]` falla (`KeyError`). Por eso existe `.get(clave, 0)`: devuelve el valor o un valor por defecto sin romper. No es para memorizar: es la solución a este fallo.

```python
# EJECUTA y compara los dos comportamientos.
try:
    stock["boli"]
except KeyError as error:
    print("KeyError:", error)
print(stock.get("boli", 0))  # sin fallo: vale 0
assert stock.get("boli", 0) == 0
assert "boli" not in stock  # .get no crea la clave
assert stock.get("lapiz", 0) == 3
print("OK: [] exige que exista; .get() ofrece salida segura")
```

## 10. Contar con `.get(clave, 0)` (ejemplo real de B2)

Contar es: si la palabra ya está, suma 1; si no, empieza en 0. `.get(palabra, 0)` evita preguntar si existe. El `for` ya lo conoces de B1.

```python
# EJECUTA. PREDICE cuánto valdrá conteo["python"].
conteo = {}
for palabra in ["python", "datos", "python"]:
    conteo[palabra] = conteo.get(palabra, 0) + 1
print(conteo)
assert conteo == {"python": 2, "datos": 1}
assert conteo["python"] == 2
print("OK: cada clave apunta a su contador")
# CAMBIA (objetivo + pista): cuenta ["rojo", "azul", "rojo", "rojo"]. Predice el dict.
```

## 11. Recorrer pares con `.items()`

`.items()` entrega **parejas** clave-valor; el `for` las recoge en dos variables. Úsalo para revisar o filtrar el diccionario.

```python
# EJECUTA. PREDICE cuántas líneas salen y qué se guarda en escasos.
tienda = {"lapiz": 4, "cuaderno": 2, "goma": 5}
for producto, cantidad in tienda.items():
    print(producto, cantidad)
escasos = []
for producto, cantidad in tienda.items():
    if cantidad <= 2:
        escasos.append(producto)
print(escasos)
assert escasos == ["cuaderno"]
print("OK: cada vuelta trabaja con una clave y su valor")
```

## 12. Comprehensions: primero el `for`, luego la versión corta

A veces transformas o filtras una lista. Hazlo primero con `for` (siempre se entiende). Después mira la comprehension equivalente. **Regla: si la versión corta cuesta más entenderla, usa el `for`.**

```python
# Paso 1: versión con for (la que ya entiendes de B1).
numeros = [1, 2, 3, 4, 5, 6]
pares_for = []
for n in numeros:
    if n % 2 == 0:
        pares_for.append(n)
print(pares_for)  # PREDICE
```

```python
# Paso 2: la MISMA idea en una línea. ¿Qué parte transforma? ¿Cuál filtra?
pares = [n for n in numeros if n % 2 == 0]
cuadrados = [n * n for n in pares]
print(pares)
print(cuadrados)
assert pares == pares_for == [2, 4, 6]
assert cuadrados == [4, 16, 36]
assert numeros == [1, 2, 3, 4, 5, 6]  # la original no cambia
print("OK: filtrar y transformar sin romper el original")
```

## 13. Analizador de frases (junta todo lo guiado)

Observa el patrón: normalizar → separar → total + únicas + frecuencias. Es el caso real “Analizador de frase” de B2. Predice totales y únicas ANTES de ejecutar.

```python
# EJECUTA. PREDICE: ¿total? ¿únicas? ¿frecuencia de "python"?
frase_analisis = "Python datos python codigo datos"
palabras_analisis = frase_analisis.lower().split()
total = len(palabras_analisis)
unicas_analisis = set(palabras_analisis)
frecuencias = {}
for palabra in palabras_analisis:
    frecuencias[palabra] = frecuencias.get(palabra, 0) + 1
print("Total:", total)
print("Unicas:", unicas_analisis)
print("Frecuencias:", frecuencias)
mas_repetida = ""
veces_max = 0
for palabra in frecuencias:
    if frecuencias[palabra] > veces_max:
        mas_repetida = palabra
        veces_max = frecuencias[palabra]
print(f"La más repetida es '{mas_repetida}' con {veces_max}")
assert total == 5
assert len(unicas_analisis) == 3
assert frecuencias == {"python": 2, "datos": 2, "codigo": 1}
assert mas_repetida in ("python", "datos")
print("OK: total, únicas y frecuencias")
```
