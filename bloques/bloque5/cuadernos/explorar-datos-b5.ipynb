```python
import numpy as np
import pandas as pd

df = pd.read_csv("got_1.csv")
print(type(df))
print(df.head())
print(df.columns.tolist())
```

```python
# PREDICE: ¿cuántas filas y columnas verás? Ejecuta y comprueba.
print(df[["Personaje", "Casa", "Estado"]])
```

```python
df_personajes = df.set_index("Personaje")
print(df_personajes.loc["Sansa Stark"])
print(df_personajes.iloc[0])
# loc = por etiqueta (nombre); iloc = por posición (número).
```

```python
# Filtro: vivos de Stark o Lannister. OJO: con Series se usa & (no and)
# y cada condición va entre paréntesis.
filtro = (df["Casa"].isin(["Stark", "Lannister"])) & (df["Estado"] == "Vivo")
print(df[filtro])
```

## Un ausente no es un cero

Bran Stark no tiene popularidad: Pandas guarda `NaN`. La media ignora ese hueco; si lo contaras como cero, la media mentiría.

```python
print(df["Popularidad"].mean())
print(np.nan)  # marca de "dato ausente", no vale cero
```

```python
# MINI-RETO del cuaderno: el dataset del proyecto usa las mismas ideas.
# (copia local para que el cuaderno sea autónomo)
hw = pd.read_csv("datos_hardware.csv")
hw = hw.set_index("id")
disponibles = hw[(hw["activo"] == "si") & (hw["stock"] > 0)]
print(disponibles)
precios = np.array(disponibles["precio"])
print("precio medio:", round(float(precios.mean()), 2))
```
