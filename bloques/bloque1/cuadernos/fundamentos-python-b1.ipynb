## 7. Calcular: `+ - * / // % **`

```python
# PREDICE cada línea antes de ejecutar.
print(7 + 3)   # suma
print(7 / 2)   # división: siempre decimal -> 3.5
print(7 // 2)  # cociente entero -> 3
print(7 % 2)   # resto -> 1
print(2 ** 3)  # potencia -> 8
```

## 11. Repetir con `for` y `range()`

**Contar y sumar:** un acumulador es una variable que crece en cada vuelta. Se crea ANTES del bucle y se actualiza DENTRO. Si la actualizas fuera, no cuenta.

```python
total = 0
for numero in range(1, 6):
    total = total + numero
print(total)  # predice: 15
assert total == 15
# CAMBIA: suma del 1 al 10. Predice (debe dar 55) y comprueba.
```

```python
# MINI-RETO (práctica real 04): ¿cuántos múltiplos de 3 hay del 1 al 30?
contador = 0
for numero in range(1, 31):
    if numero % 3 == 0:
        contador = contador + 1
print(contador)
assert contador == 10
print("OK: hay 10 múltiplos de 3 entre 1 y 30")
# MINI-RETO: cuenta múltiplos de 5 entre 1 y 100 (debe dar 20).
```

## 12. Repetir sin saber cuántas veces: `while`

`for` sirve cuando sabes cuántas vueltas (un `range`). `while` repite MIENTRAS una condición sea verdadera, para cuando no sabes cuántas vueltas harán falta. Algo tiene que cambiar dentro, o no termina nunca.

```python
intentos = 0
while intentos < 3:
    intentos = intentos + 1
    print("intento", intentos)
# PREDICE: ¿cuántas líneas salen?
# Pregunta clave: ¿qué pasaría si borraras la línea intentos = intentos + 1?
```

**ERROR ÚTIL:** si nada cambia dentro del `while`, la condición nunca se vuelve falsa y el programa no termina (bucle infinito). Por eso esto de abajo NO debes ejecutarlo: `while 1 < 2:` sin nada que avance. Regla: todo `while` necesita algo que avance hacia el final.

## 13. `break` y `continue` (pequeños)

`break` sale del bucle del todo. `continue` salta el resto de la vuelta y pasa a la siguiente. Solo úsalos en bucles pequeños que entiendas.

```python
for n in range(1, 10):
    if n == 4:
        break
    print(n)
# PREDICE: se muestra 1, 2, 3 y se acabó (el 4 rompe el bucle).
```

```python
# Práctica real 04: continue salta lo que no interesa.
contador = 0
for numero in range(1, 31):
    if numero % 3 != 0:
        continue
    contador = contador + 1
print(contador)
assert contador == 10
print("OK: continue salta los que no son múltiplos")
```

**¿Cuándo `while` de verdad?** Cuando dependes de la persona: pedir números hasta que escriba 0. Con `for` no puedes, porque no sabes cuántos pedirá.

```python
# MINI-RETO (reto real de la práctica 04): pide números hasta el 0
# y cuenta cuántos fueron. Ejemplo de sesión: 5, 8, 0 -> 2 números.
cantidad = 0
numero = int(input("Número (0 para terminar): "))
while numero != 0:
    cantidad = cantidad + 1
    numero = int(input("Número (0 para terminar): "))
print("Has introducido", cantidad, "números")
```
