## 4. `return` vs `print`: el concepto central

`print()` **muestra** en pantalla. `return` **devuelve** un valor para seguir trabajando con él. Son trabajos distintos.

⚠️ **PREDICE antes de ejecutar:** ¿qué guarda `resultado`? ¿Qué imprime el último `print`?

````python
# PREDICE y luego EJECUTA.
def mostrar_doble(n):
    print(n * 2)

def devolver_doble(n):
    return n * 2

resultado = mostrar_doble(5)   # PREDICE: ¿qué contiene 'resultado'?
print("resultado vale:", resultado)
print("tipo:", type(resultado))

valor = devolver_doble(5)
print(valor + 1)               # PREDICE: ¿qué imprime?
# COMPRUEBA
assert resultado is None
assert valor == 10
assert valor + 1 == 11
print("OK: print muestra; return entrega un valor reutilizable")
````

Una función **sin `return`** devuelve siempre `None`. Por eso `resultado` vale `None`: `mostrar_doble` imprime 10 en pantalla, pero no entrega nada para guardar.

## 6. Ámbito local: lo creado dentro no sale fuera

Una variable creada **dentro** de una función es **local**: solo existe ahí dentro. Fuera manda la variable exterior, aunque se llame igual.

````python
# PREDICE los dos prints antes de ejecutar.
usuario = "Invitado"

def cambiar_nombre():
    usuario = "Admin"   # esta 'usuario' es local: nace y muere aquí
    return usuario

print(cambiar_nombre())  # PREDICE
print(usuario)           # PREDICE: ¿cambió la exterior?
# COMPRUEBA
assert cambiar_nombre() == "Admin"
assert usuario == "Invitado"
print("OK: la local no toca la exterior")
````

````python
# COMPRUEBA el límite: usar la local fuera da NameError.
def crear_ticket():
    codigo_ticket = "T-001"
    return codigo_ticket

crear_ticket()
try:
    print(codigo_ticket)
except NameError as error:
    print("NameError:", error)
print("Lección: si la necesitas fuera, devuélvela con return.")
````

**`global`, solo como caso controlado.** Hay una excepción: si una función *debe* modificar una variable exterior (por ejemplo un contador), se declara con `global`. No es la primera opción: lo normal es devolver valores con `return`. Úsalo solo cuando el diseño lo pida y de forma visible.

````python
# EJECUTA. Contador controlado con 'global' y límite de 5 visitas.
visitas = 0

def registrar_visita():
    global visitas
    if visitas < 5:
        visitas = visitas + 1
    return visitas

print(registrar_visita())
print(registrar_visita())
# COMPRUEBA
assert visitas == 2
print("OK: 'global' solo en este experimento de contador")
# MINI-RETO 5 (edita la celda anterior): cambia el límite de 5 a 3 y comprueba
# que la cuarta llamada ya no suma. Pista: solo toca el número del 'if'.
````

## 8. Error real y traceback (controlado)

Provocamos un error a propósito, **atrapado** para poder leerlo con calma. Todo error trae tres datos: **tipo** (`TypeError`), **mensaje** y **línea donde ocurrió** (en VS Code el traceback la marca; aquí la leemos con `type` y el mensaje).

````python
# EJECUTA. Pasamos texto donde se esperaba número, como haría un dato mal leído.
def calcular_precio_final_local(precio, tasa_impuesto):
    if not isinstance(precio, (int, float)):
        raise TypeError("precio debe ser numerico")
    return round(precio + precio * tasa_impuesto, 2)

try:
    calcular_precio_final_local("cien", 0.21)
except TypeError as error:
    print("Tipo de error:", type(error).__name__)
    print("Mensaje:", error)
# COMPRUEBA que sabes identificarlo
try:
    calcular_precio_final_local("cien", 0.21)
    tipo_detectado = "ninguno"
except TypeError:
    tipo_detectado = "TypeError"
assert tipo_detectado == "TypeError"
print("OK: lees tipo y mensaje antes de corregir")
````

## 9. `try` / `except`: primero el error, después la red

Caso útil: convertir texto en número. Si el texto no es numérico, `float()` lanza `ValueError`. Lo capturamos **por su nombre** y el programa sigue vivo.

````python
# EJECUTA con texto='abc' y luego cambia a '12.5'.
def convertir_precio(texto):
    """Convierte texto a numero decimal."""
    return float(texto)

for texto in ["abc", "12.5"]:
    try:
        precio = convertir_precio(texto)
        print("Convertido:", precio)
    except ValueError:
        print("Precio no valido:", texto)
# COMPRUEBA
assert convertir_precio("12.5") == 12.5
print("OK: el programa no se detiene y ayuda a corregir")
````

⚠️ **No tapes todos los errores con `except:` a secas.** Un `except` sin nombre captura hasta tus despistes (como una variable mal escrita) y te deja sin pistas. Captura el error que **esperas**.

````python
# ERROR ÚTIL: el 'except' desnudo esconde un despiste (precioss con doble s).
def total_cesta(precios):
    return sum(precios)

cesta = [10, 20, 30]
try:
    total = total_cesta(precioss)
    print(total)
except:
    print("Algo fallo (¿el qué? no lo sabes)")

# Así sí: el error esperado se nombra y el despiste queda visible.
try:
    total = total_cesta(precioss)
except NameError as error:
    print("NameError visible:", error)
    mensaje = str(error)
assert "precioss" in mensaje
assert total_cesta(cesta) == 60
print("OK: except con nombre, nunca a ciegas")
````

## 11. `with open(...)`: abrir, usar y cierre automático

`with open(...)` abre el archivo, lo usas dentro del bloque y **se cierra solo** al salir, incluso si algo falla. Modos: `"w"` escribe (borra lo anterior), `"r"` lee, `"a"` añade al final.

````python
# EJECUTA. Escribimos una nota y la leemos de vuelta.
with open("nota_b3.txt", "w", encoding="utf-8") as archivo:
    archivo.write("Primera nota de B3")

with open("nota_b3.txt", "r", encoding="utf-8") as archivo:
    contenido = archivo.read()

print(contenido)
# COMPRUEBA
assert contenido == "Primera nota de B3"
print("OK: escribes, lees y nada queda abierto")
````

## 12. TXT / CSV / JSON: una pequeña comparación

No son tres cursos; es una tabla:

| Formato | Sirve para | En Python |
|---|---|---|
| TXT | texto libre (notas, logs) | `write` / `read` |
| CSV | tablas (filas y columnas) | módulo `csv` |
| JSON | datos estructurados (diccionarios) | módulo `json` |

SAMI-Lite usa los tres igual: TXT para el log de errores, JSON para la configuración y CSV para las transacciones. Priorizamos JSON y CSV porque el proyecto los necesita.

````python
# EJECUTA. JSON: un diccionario que viaja al archivo y vuelve como diccionario.
import json

config = {"umbral_alerta": 100.0, "iva_aplicado": 0.21}
with open("config_demo_b3.json", "w", encoding="utf-8") as archivo:
    json.dump(config, archivo, indent=2)

with open("config_demo_b3.json", "r", encoding="utf-8") as archivo:
    cargada = json.load(archivo)

print(cargada["iva_aplicado"])
# COMPRUEBA
assert isinstance(cargada, dict)
assert cargada["iva_aplicado"] == 0.21
print("OK: JSON guarda y recupera la estructura")
````

````python
# EJECUTA. CSV: cabecera + filas, tabla sencilla.
import csv

with open("transacciones_demo_b3.csv", "w", newline="", encoding="utf-8") as archivo:
    escritor = csv.writer(archivo)
    escritor.writerow(["producto", "precio_base", "precio_final", "estado"])
    escritor.writerow(["Teclado", 50, 60.5, "PRECIO_NORMAL"])
    escritor.writerow(["Monitor", 100, 121.0, "ALERTA_PRECIO_ELEVADO"])

with open("transacciones_demo_b3.csv", "r", encoding="utf-8") as archivo:
    filas = list(csv.reader(archivo))

print(filas)
# COMPRUEBA: 1 cabecera + 2 datos = 3 filas; PREDICE antes de ejecutar.
assert len(filas) == 3
assert filas[0] == ["producto", "precio_base", "precio_final", "estado"]
print("OK: CSV con columnas claras y filas consistentes")
````

### MINI-RETO 7 — objetivo + pista

**Objetivo:** crea `cargar_preferencias(ruta)` que devuelva el diccionario guardado; si el archivo aún no existe, lo crea con valores por defecto.

**Pistas:** `try` + `json.load`; `except FileNotFoundError` + `json.dump` de `{"tema": "claro"}`; es el mismo truco que SAMI-Lite usa para su `config.json`.

````python
# MINI-RETO 7: base que ya ejecuta; escríbela tú primero.
import json

def cargar_preferencias(ruta):
    """Carga preferencias o crea valores por defecto si no existen."""
    por_defecto = {"tema": "claro"}
    try:
        with open(ruta, "r", encoding="utf-8") as archivo:
            return json.load(archivo)
    except FileNotFoundError:
        with open(ruta, "w", encoding="utf-8") as archivo:
            json.dump(por_defecto, archivo, indent=2)
        return por_defecto

preferencias = cargar_preferencias("preferencias_demo_b3.json")
print(preferencias)
assert isinstance(preferencias, dict)
assert cargar_preferencias("preferencias_demo_b3.json") == preferencias
print("OK: dominas el patrón 'cargar o crear'")
````

# Parte 4 — De notebook a `.py`: el programa se reparte

Hasta aquí vivías en celdas:

```text
Notebook:  celda → celda → celda   (todo en un archivo)
```

Un programa real se reparte en archivos, cada uno con una responsabilidad:

```text
Programa:  main.py  →  analizador.py  →  persistencia.py
           (coordina)  (calcula)        (guarda y carga)
```

**Concepto:** un **módulo** es simplemente otro archivo Python que reutilizamos con `import`. Solo aprenderemos las dos formas que el proyecto necesita:

- `from analizador import calcular_precio_final` → traigo una función concreta.
- `import persistencia` → traigo el módulo y uso `persistencia.cargar_configuracion(...)`.
